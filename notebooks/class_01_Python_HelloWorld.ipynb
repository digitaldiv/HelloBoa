### Variable naming
A variable can have a short name (like x and y) or a more descriptive name (age, carname, total_volume). 

**Rules for Python variables:**

- A variable name must start with a letter or the underscore character
- A variable name cannot start with a number
- A variable name can only contain alpha-numeric characters and underscores (A-z, 0-9, and _ )
- Variable names are case-sensitive (age, Age and AGE are three different variables)
- A variable name cannot be any of the Python keywords.

```python
# Legal variable names
myvar = "John"
my_var = "John"
_my_var = "John"  # Underscore 💪🏼 has a special place in Python 
myVar = "John"
MYVAR = "John"
myvar2 = "John"

```

```python
# Illegal variable names
# that cuase errors
2myvar = "John"
my-var = "John"
my var = "John"
```

### Bitwise Operators

|Operator | Operation | Example |
|---------|-----------------------|---------|
| & | AND : Sets each bit to 1 if both bits are 1	| x & y |
| &#124; | OR : Sets each bit to 1 if one of two bits is 1	| x &#124; y |
|^	| XOR :	Sets each bit to 1 if only one of two bits is 1| x ^ y	|

| <<|	Zero fill left shift	Shift left by pushing zeros in from the right and let the leftmost bits fall off	| x << 2	|

# Practice Problems: Hello Boa 

Here are some of the practice problems that will help reinforce concepts like variables, data types, operators, input/output, and Python syntax basics.


## Basic Output and Variables

1. **Hello Python**: Write a program that prints your name and your favorite programming language on separate lines.

2. **Variable Types**: Create four variables with different data types (integer, float, string, and boolean) and print each variable along with its data type.

3. **Multiple Assignment**: Use multiple assignment to create three variables (x, y, z) all with the value of 100, then print each variable.

## String Operations

4. **Name Formatter**: Ask the user for their first name and last name as separate inputs, then print their full name in the format "Last name, First name".

5. **String Repetition**: Ask the user for a short word and a number. Print the word repeated that number of times, all on one line.

6. **String Concatenation**: Create two string variables and concatenate them with a space in between, then print the result.

## Numeric Operations

7. **Simple Calculator**: Ask the user for two numbers and print the results of adding, subtracting, multiplying, and dividing them.

8. **Temperature Converter**: Write a program that converts a temperature from Celsius to Fahrenheit. The formula is: F = (C × 9/5) + 32

9. **Modulus Practice**: Ask the user for two integers and print the remainder when the first number is divided by the second.

## Type Conversion

10. **Age Calculator**: Ask the user for their birth year and calculate their approximate age. Remember to convert the input to an integer!

11. **Float to Integer**: Ask the user for a decimal number, convert it to an integer (removing the decimal portion), and print both the original and converted values.

## Python Syntax
